####  ✅ Selected numerical features and computed their correlation matrix to analyze relationships between quantitative variables.

#### This matrix presents the Pearson correlation coefficients between different features  in a dataset.
##### The coefficients range from -1 to 1, indicating how strongly features are related.
##### 1 or -1: Perfect correlation (positive or negative).
 #####       0: No correlation.
 ##### Positive Values: Indicate that as one feature increases, the other tends to increase as well.
##### Negative Values: Indicate that as one feature increases, the other tends to decrease.


```python
numerical_data = data.select_dtypes(include=['number'])
numerical_data.corr()
```

#### ✅ Created a scatter plot to visualize the relationship between total working years and monthly income.
#### 📊 Key Insight
**There is a clear positive relationship between total working years and monthly income — employees with more experience generally earn higher salaries.**

```python
plt.figure(figsize=(8,5))
plt.scatter(data['TotalWorkingYears'], data['MonthlyIncome'])
plt.xlabel("Total Working Years")
plt.ylabel("Monthly Income")
plt.title("Monthly Income vs Total Working Years")
```

#### ✅ Sorted the dataset by total working years and plotted a line chart to observe the trend of monthly income with increasing experience.
#### 📊 Key Insight
**Monthly income generally increases as total working years increase, showing a clear upward income trend with experience.**

```python
data_sorted = data.sort_values(by='TotalWorkingYears')

plt.figure(figsize=(8,5))
plt.plot(data_sorted['TotalWorkingYears'], data_sorted['MonthlyIncome'])
plt.xlabel("Total Working Years")
plt.ylabel("Monthly Income")
plt.title("Trend of Monthly Income with Experience")
plt.show()
```

####  ✅ Created a correlation heatmap to visualize relationships between numerical features in the dataset.
### 📊 Key Insights 
**Strong Positive Relationships**_
**TotalWorkingYears ↔ MonthlyIncome**
**YearsAtCompany ↔ YearsInCurrentRole**
**YearsAtCompany ↔ YearsSinceLastPromotion**

#### 📊 Key Combined Insights From These Plots
🔹 Age Group

       Highest attrition appears in 25–35 and 30–40 age groups
        👉 Early-career and mid-career employees are more likely to leave.
🔹 Business Travel

       Employees who travel frequently show higher attrition.
        👉 Travel-related fatigue may impact retention.
🔹 Department

       Sales and R&D show higher attrition than HR.
       👉 Some departments need focused retention strategies.
🔹 Education Field

      Technical fields (Life Sciences, Medical, Technical Degree) show higher counts.
       👉 Competitive job markets may drive turnover.
🔹 Gender

      Slightly higher attrition among males, but difference is small.
       👉 Gender is not a strong attrition driver.
🔹 Job Role

      Sales Executive, Research Scientist, Laboratory Technician show higher attrition.
       👉 Frontline and technical roles are more vulnerable.
🔹 Marital Status

      Single employees show higher attrition.
       👉 Life stage influences job stability.

```python
selected_cols = [
    'Age-Group', 'BusinessTravel', 'Department',
    'EducationField', 'Gender', 'JobRole', 'MaritalStatus'
]

for i, predictor in enumerate(selected_cols):
    plt.figure(figsize=(20,4))
    sns.countplot(data=data, x=predictor, hue='Attrition')


```
